```python
from datasets import load_dataset

dataset = load_dataset("csv", data_files="data/urbandict-word-defs.csv", engine='python', on_bad_lines='skip')["train"]

def add_text(example):
    example["text"] = f"{example['word']}: {example['definition']}"
    return example

dataset = dataset.map(add_text)
dataset = dataset.train_test_split(test_size=0.05)

print(dataset)
print(dataset["train"][:1])
```

```python
# Tokenization

char_set = set()
def add_to_char_set(example):
    chars = list(example["text"])
    char_set.update(chars)

dataset.map(add_to_char_set)
```

```python
print(f"Unique characters:\n{sorted(list(char_set))}")
print("\n" in char_set)
```

```python
import tiktoken
enc = tiktoken.get_encoding("gpt2")
toks = enc.encode("word: silly misspelllllled definition")
print(toks)
print(enc.decode(toks))
```
